Question 11 : Détection par IQR et traitement (2 points)
1. [/1.5] Détection par écart interquartile de Quantity :
○ Calculez l'IQR (InterQuartile Range) : IQR = Q3 - Q1.
○ Définissez les bornes pour les outliers : limite_inf = Q_1 - 1.5 * IQR et limite_sup =
Q_3 + 1.5 * IQR.
○ Comptez le nombre d'outliers détectés par cette méthode dans la colonne Quantity.
2. [/0.5] Parmi les trois approches suivantes pour traiter les outliers (suppression, imputation
par la médiane/mode, justification sans modification), laquelle choisiriez-vous et pourquoi ?

```python
# Détection par écart interquartile de Quantity :
# ○ Calculez l'IQR (InterQuartile Range) : IQR = Q3 - Q1.
Q1 = df_sales_clean['Quantity'].quantile(0.25)
Q3 = df_sales_clean['Quantity'].quantile(0.75)
IQR = Q3-Q1

#  Définissez les bornes pour les outliers : limite_inf = Q_1 - 1.5 * IQR et limite_sup = Q_3 + 1.5 * IQR.
limite_inf = Q1 - 1.5 * IQR
limite_sup = Q3 + 1.5 * IQR

# Comptez le nombre d'outliers détectés par cette méthode dans la colonne Quantity.
outliers_iqr = df_sales_clean[(df_sales_clean['Quantity'] < limite_inf) | (df_sales_clean['Quantity'] > limite_sup)]
print(f"Outliers détectés : {len(outliers_iqr)}")
```

On choisit la justification sans modification parce dans une entreprise commeriale, les grosses commandes sont réelles et importantes pour le CA. Si on les supprime, on peut fausser les calculs comptables et donc faussées plus gravement nos analyses.
